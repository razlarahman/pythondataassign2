```python
import seaborn as sns
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["payment"] = df["payment"].fillna(df["payment"].mode()[0])
```

```python
df["pickup_zone"] = df["pickup_zone"].fillna(df["pickup_zone"].mode()[0])
```

```python
df["dropoff_zone"] = df["dropoff_zone"].fillna(df["dropoff_zone"].mode()[0])
```

```python
df["pickup_borough"] = df["pickup_borough"].fillna(df["pickup_borough"].mode()[0])
```

```python
df["dropoff_borough"] = df["dropoff_borough"].fillna(df["dropoff_borough"].mode()[0])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

```

```python
plt.plot(df["pickup"], df["fare"], marker="o")
plt.xlabel("pickup date and time")
plt.ylabel("fare")
plt.title("pickup vs fare")
plt.xticks(rotation=45)
plt.show()
```

```python
data = df.groupby("pickup_borough")["fare"].sum()
print(data)
data.plot(kind="bar", color="skyblue", edgecolor="black")
plt.xlabel("pickup_borough.")
plt.ylabel("fare")
plt.title("fare vs pickup borough")
```

```python
data = df["payment"].value_counts()
print(data)
plt.pie(data,autopct="%1.1f%%")
plt.title("payment")

plt.legend(data.index)
plt.show()

```

```python
plt.hist(df["distance"],bins=10)
plt.title("distance")
plt.show()
```

```python
data = df.groupby("pickup_borough")["tip"].unique()
print(data)
plt.boxplot(data,labels=df["pickup_borough"].unique())
plt.xlabel("Pickup Borough")
plt.ylabel("Tip")
plt.title("Tip Amount by Pickup Borough")
plt.show()
```

```python
#seaborn
sns.countplot(x="pickup_borough",data=df)
plt.title("pickup borough")
plt.show()
```

```python
sns.scatterplot(x="distance", y="fare", hue="pickup_borough", data=df)
plt.title('Relationship Between Distance and Fare by Pickup Borough')
plt.show()
```

```python
corr = df.select_dtypes("number").corr()
sns.heatmap(corr,annot=True)
plt.show()
```

```python

sns.pairplot(df, hue="pickup_zone")
plt.show()
```

```python
sns.violinplot(x="payment", y="fare", data=df)
plt.show()
```
